```python
import pandas as pd
import numpy as np
from datetime import timedelta

import scripts.stock_plots as stock_plots
from scripts.preparation import download_data
```

# Broad market

```python
ticker_list = ["^GSPC", "^SPXEW", "^DJI", "^RUT", "^IXIC"]
market_explain = ["sp500 broad market",
                  "equal sp500, more on smaller cap",
                  "leaders in sectors",
                  "russel small caps",
                  "nas on tech"]
ticker_object = download_data(ticker_list)

for ticker, explain in zip(ticker_list, market_explain):
    stock_plot = stock_plots.PlotInfo(ticker_object.tickers[ticker], f"{ticker} {explain}", "130d")
    # candle = stock_plot.generate_quick_analysis(days=30)
    candle = stock_plot.generate_candle_plot_no_op()
    candle.show()
```

# Holdings analysis


```python
watchlist = pd.read_excel("watchlist.xlsx")
holdings = watchlist[["Op Ticker", "Contract", "Type", "Open", "Close"]].dropna().astype(str)
print(len(holdings))
```

```python
ticker_list = holdings["Op Ticker"]
ticker_object = download_data(ticker_list)

for idx, row in holdings.iterrows():
    ticker = row["Op Ticker"]
    title = f"{ticker} {row['Type']}"
    strikes = [float(i) for i in row["Contract"].split()]

    stock_plot = stock_plots.PlotInfo(ticker_object.tickers[ticker], title, "100d")
    # candle = stock_plot.generate_quick_analysis(days=20)
    candle = stock_plot.generate_candle_plot_no_op()

    # add the strike prices
    candle.add_hline(y=np.min(strikes), line_dash="dot", row=1, col=1, line_color="black", line_width=0.7)
    candle.add_hline(y=np.max(strikes), line_dash="dot", row=1, col=1, line_color="black", line_width=0.7)
    candle.add_hline(y=np.mean(strikes), line_dash="dot", row=1, col=1, line_color="black", line_width=0.7)

    candle.add_vline(x=row["Open"], line_dash="dot", row=1, col=1, line_color="black", line_width=1)
    candle.add_vline(x=row["Close"], line_dash="dot", row=1, col=1, line_color="black", line_width=1)

    candle.show()


```

# The watchlist

```python
ticker_list = watchlist["Watching"]
ticker_object = download_data(ticker_list)

for ticker in ticker_list:
    if ticker in list(holdings["Op Ticker"]):
        continue
    stock_plot = stock_plots.PlotInfo(ticker_object.tickers[ticker], ticker, "100d")
    # candle = stock_plot.generate_quick_analysis(days=50)
    candle = stock_plot.generate_candle_plot_no_op()

    candle.show()
```
